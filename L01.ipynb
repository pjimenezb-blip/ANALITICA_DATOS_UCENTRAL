## ¿Qué es un número?
* Una secuencia de dígitos
* Considere el número (decimal) 1979:
    * 1979 = 1x10<sup>3</sup> + 9x10<sup>2</sup> + 7x10<sup>1</sup>  + 9x10<sup>0</sup>
    * Esto se conoce como representación en base 10


## Representación binaria
* Las computadoras emplean una representación en base 2
* Dígito binario = "bit" ∈ {0,1}

* Considere un número de 4 bits: 
* 1010<sub>2</sub>  = 1x2<sup>3</sup> + 0x2<sup>2</sup>+ 1x2<sup>1</sup> + 0x2<sup>0</sup>  = 10<sub>10</sub>

## Igualdad versus asignación
* Igualdad
    * x=2
    * "x" es una variable equivalente a "dos"
    * 2=x es una ecuación válida y equivalente
* Asignación
    * x≔2
    * "x" es una variable cuyo valor se está estableciendo en "dos"
    * 2≔x no es una afirmación válida

## ¿Qué hacen las computadoras?

* A nivel físico, las computadoras manipulan pequeños voltajes eléctricos
* Las computadoras ejecutan secuencias de instrucciones
* Las computadoras suman, restan, multiplican y dividen números binarios
* Las computadoras comparan voltajes (números) entre sí
    * Lo que ocurre después depende del resultado de esta comparación

* Todo el software, desde el que controla tu refrigerador hasta el último videojuego Call of Duty, se reduce a un conjunto relativamente pequeño de operaciones sobre números binarios


## Ley de Moore
* El número de transistores en un microchip se duplica aproximadamente cada dos años
* El costo del microchip se reduce a la mitad cada dos años
* La ley de Moore se aplica a la memoria del programa (RAM), al almacenamiento secundario y a la velocidad del procesador

## Vamos a crear una variable para almacenar el tiempo

````python
time = np.linspace(0,100,101) 
time
````

````python
n_transistors = 2**(time/2)
n_transistors
````
